# Day 3 – Data cleaning and validation

Do not remove rows just because they contain missing values. Inspect issues, understand their meaning and document cleaning decisions first.

```python
if "df_raw" in globals():
    df_work = df_raw.copy()
    print("Working copy created:", df_work.shape)
else:
    print("Load the dataset first.")
```

```python
# 1. Missing-value report
if "df_work" in globals():
    missing_report = pd.DataFrame({
        "missing_count": df_work.isna().sum(),
        "missing_percentage": (df_work.isna().mean() * 100).round(2)
    }).sort_values("missing_count", ascending=False)
    display(missing_report)
```

```python
# 2. Duplicate report
if "df_work" in globals():
    print("Exact duplicate rows:", int(df_work.duplicated().sum()))
    if "booking_id" in df_work.columns:
        print("Missing booking IDs:", int(df_work["booking_id"].isna().sum()))
        print("Duplicate non-null booking IDs:",
              int(df_work["booking_id"].dropna().duplicated().sum()))
```

```python
# 3. Validation checks for important fields
if "df_work" in globals():
    checks = {}

    if "is_canceled" in df_work.columns:
        checks["invalid_is_canceled"] = int(
            (~df_work["is_canceled"].isin([0, 1]) & df_work["is_canceled"].notna()).sum()
        )

    if "lead_time" in df_work.columns:
        lead = pd.to_numeric(df_work["lead_time"], errors="coerce")
        checks["non_numeric_lead_time"] = int((df_work["lead_time"].notna() & lead.isna()).sum())
        checks["negative_lead_time"] = int((lead < 0).sum())

    if "average_daily_rate" in df_work.columns:
        adr = pd.to_numeric(df_work["average_daily_rate"], errors="coerce")
        checks["non_numeric_adr"] = int((df_work["average_daily_rate"].notna() & adr.isna()).sum())
        checks["negative_adr"] = int((adr < 0).sum())

    display(pd.Series(checks, name="issue_count").to_frame())
```

```python
# 4. Create a cleaned working copy
if "df_work" in globals():
    df_clean = df_work.copy()

    # Remove exact duplicate rows only.
    df_clean = df_clean.drop_duplicates()

    # Safely convert selected columns to numeric when present.
    numeric_candidates = [
        "is_canceled", "lead_time", "average_daily_rate", "adults",
        "children", "stays_in_weekend_nights", "stays_in_week_nights",
        "booking_changes"
    ]
    for column in numeric_candidates:
        if column in df_clean.columns:
            df_clean[column] = pd.to_numeric(df_clean[column], errors="coerce")

    print("Original shape:", df_work.shape)
    print("After exact duplicate removal:", df_clean.shape)
    display(df_clean.head())
```

```python
# 5. Missing children example – apply only if the data definition confirms
# that a missing children value means zero.
if "df_clean" in globals() and "children" in df_clean.columns:
    print("Missing children values:", int(df_clean["children"].isna().sum()))
    # After confirming with your team, you may apply:
    # df_clean["children"] = df_clean["children"].fillna(0)
```

```python
# 6. Save the cleaned working copy
if "df_clean" in globals():
    output_dir = Path("data/cleaned")
    output_dir.mkdir(parents=True, exist_ok=True)
    cleaned_path = output_dir / "cleaned_hotel_bookings_w1.csv"
    df_clean.to_csv(cleaned_path, index=False)
    print("Saved:", cleaned_path)
    print("Saved shape:", df_clean.shape)
```

### Important cleaning notes
- Duplicate booking IDs and exact duplicate rows are different issues. Investigate repeated IDs before deleting records.
- Do not replace every missing value with zero. Zero has a business meaning.
- Negative ADR and negative lead time should be flagged and investigated, not silently removed.
- Confirm `is_canceled` coding before calculating cancellation rates.
- Keep the raw dataset unchanged and save cleaned data separately.

# Day 4 – Statistics and basic hotel business summaries

Use the cleaned working copy where available. These calculations describe the data; they do not prove why a cancellation happened.

```python
if "df_clean" in globals():
    df_analysis = df_clean.copy()
elif "df_raw" in globals():
    df_analysis = df_raw.copy()
else:
    df_analysis = None
    print("Load the dataset first.")
```

```python
# 1. Descriptive statistics for numerical columns
if df_analysis is not None:
    numeric_df = df_analysis.select_dtypes(include=np.number)
    display(numeric_df.describe().T)
```

```python
# 2. Mean, median, mode, min, max and standard deviation
if df_analysis is not None:
    columns_to_summarize = [
        c for c in ["lead_time", "average_daily_rate", "stays_in_weekend_nights",
                    "stays_in_week_nights", "booking_changes"]
        if c in df_analysis.columns
    ]

    rows = []
    for column in columns_to_summarize:
        values = pd.to_numeric(df_analysis[column], errors="coerce").dropna()
        if not values.empty:
            modes = values.mode()
            rows.append({
                "column": column,
                "mean": values.mean(),
                "median": values.median(),
                "mode": modes.iloc[0] if not modes.empty else np.nan,
                "minimum": values.min(),
                "maximum": values.max(),
                "standard_deviation": values.std()
            })

    display(pd.DataFrame(rows).round(2))
```

```python
# 3. Cancellation summary – calculate only if values are valid 0/1 flags
if df_analysis is not None and "is_canceled" in df_analysis.columns:
    flags = pd.to_numeric(df_analysis["is_canceled"], errors="coerce").dropna()
    print("Observed values:")
    display(flags.value_counts().sort_index())

    if not flags.empty and flags.isin([0, 1]).all():
        total = len(flags)
        cancelled = int((flags == 1).sum())
        not_cancelled = int((flags == 0).sum())
        rate = cancelled / total * 100

        print("Valid records:", total)
        print("Cancelled:", cancelled)
        print("Not cancelled:", not_cancelled)
        print("Cancellation rate:", round(rate, 2), "%")
    else:
        print("Confirm 0/1 coding before calculating cancellation rate.")
```

```python
# 4. Room-type comparison
if df_analysis is not None and {"reserved_room_type", "is_canceled"}.issubset(df_analysis.columns):
    room_summary = df_analysis.groupby("reserved_room_type")["is_canceled"].agg(
        total_records="count",
        cancellation_flag_sum="sum",
        cancellation_proportion="mean"
    )
    room_summary["cancellation_percentage"] = room_summary["cancellation_proportion"] * 100
    display(room_summary.round(2))
else:
    print("This comparison needs reserved_room_type and is_canceled.")
```

```python
# 5. Market-segment comparison
if df_analysis is not None and {"market_segment", "is_canceled"}.issubset(df_analysis.columns):
    market_summary = df_analysis.groupby("market_segment")["is_canceled"].agg(
        total_records="count",
        cancellation_flag_sum="sum",
        cancellation_proportion="mean"
    )
    market_summary["cancellation_percentage"] = market_summary["cancellation_proportion"] * 100
    display(market_summary.round(2))
else:
    print("This comparison needs market_segment and is_canceled.")
```

```python
# 6. Arrival-month comparison
if df_analysis is not None and {"arrival_month", "is_canceled"}.issubset(df_analysis.columns):
    monthly_summary = df_analysis.groupby("arrival_month")["is_canceled"].agg(
        total_records="count",
        cancellation_flag_sum="sum",
        cancellation_proportion="mean"
    )
    monthly_summary["cancellation_percentage"] = monthly_summary["cancellation_proportion"] * 100
    display(monthly_summary.round(2))
else:
    print("This comparison needs arrival_month and is_canceled.")
```

```python
# 7. Repeat-guest comparison
if df_analysis is not None and {"repeated_guest", "is_canceled"}.issubset(df_analysis.columns):
    repeat_summary = df_analysis.groupby("repeated_guest")["is_canceled"].agg(
        total_records="count",
        cancellation_flag_sum="sum",
        cancellation_proportion="mean"
    )
    repeat_summary["cancellation_percentage"] = repeat_summary["cancellation_proportion"] * 100
    display(repeat_summary.round(2))
else:
    print("This comparison needs repeated_guest and is_canceled.")
```

```python
# 8. Correlation – association, not causation
if df_analysis is not None:
    correlation_columns = [
        c for c in ["lead_time", "average_daily_rate", "booking_changes",
                    "stays_in_weekend_nights", "stays_in_week_nights", "is_canceled"]
        if c in df_analysis.columns
    ]
    if len(correlation_columns) >= 2:
        display(df_analysis[correlation_columns].corr(numeric_only=True).round(2))
```

### How to explain the statistics
- **Mean:** Arithmetic average, such as average lead time.
- **Median:** Middle value after sorting.
- **Mode:** Most frequent value.
- **Variance:** Measure of spread in the values.
- **Standard deviation:** Typical spread around the mean.
- **Correlation:** Measures association between numerical variables. It does not prove causation.
